```python
import pandas as pd
```

```python
df = pd.read_csv("kindergardens_.csv", dtype={"location": "string"})
```

```python
df.head()
```

```python
df["location"] = (
    df["location"]
    .str.strip()
    .str.replace(
        r"^POINT\s*\(\s*([-\d.]+)\s+([-\d.]+)\s*\)$",
        r"POINT(\1 \2)",
        regex=True
    )
    .astype("string")
)
```

```python
df.head()
```

```python
df.to_csv("kindergardens_points.csv", index=False)
```
